## Part 2 — Code Implementation
Complete Minimax implementation for the sticks game.

```python
# State Representation
sticks = 4

# Terminal Test
def terminal(sticks):
    return sticks == 0

# Utility Function
def utility(sticks, is_max_turn):
    if sticks == 0:
        return -1 if is_max_turn else +1

# Successor Function
def successors(sticks):
    moves = []
    if sticks >= 1:
        moves.append(sticks - 1)
    if sticks >= 2:
        moves.append(sticks - 2)
    return moves

# Minimax Algorithm
import math
def minimax(sticks, is_max_turn):
    if terminal(sticks):
        return utility(sticks, is_max_turn)
    if is_max_turn:
        best = -math.inf
        for next_state in successors(sticks):
            val = minimax(next_state, False)
            best = max(best, val)
        return best
    else:
        best = math.inf
        for next_state in successors(sticks):
            val = minimax(next_state, True)
            best = min(best, val)
        return best

# Best Move Selector
def best_move(sticks):
    best_val = -math.inf
    best_action = None
    for move in [1, 2]:
        if sticks - move >= 0:
            val = minimax(sticks - move, False)
            if val > best_val:
                best_val = val
                best_action = move
    return best_action

# Test
sticks = 4
move = best_move(sticks)
print("Best move:", move)
```

## Part 4 — Alpha–Beta Pruning (Conceptual)
Alpha–Beta pruning is an optimization of the Minimax algorithm that reduces the number of nodes evaluated in the game tree while still producing the same optimal decision. It maintains two values:
- **Alpha (α):** the best value MAX can guarantee
- **Beta (β):** the best value MIN can guarantee
If a branch cannot improve the outcome for the current player, it is pruned and ignored.

Example: If MAX has already found a move with value +1 and another branch can only produce 0 or −1, that branch is skipped because it cannot improve MAX's result. 
Benefits of Alpha–Beta pruning include reduced computation, faster decision-making, and the ability to search deeper in complex games. It is widely used in AI for games such as chess, checkers, tic-tac-toe, and strategy video games.
