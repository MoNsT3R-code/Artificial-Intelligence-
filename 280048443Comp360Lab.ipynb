```python
import random

# Parameters
POP_SIZE = 6
CHROM_LENGTH = 10
GENERATIONS = 10
MUTATION_RATE = 0.1
```

```python
# 1. Create a random individual (binary chromosome)
def create_individual():
    return [random.randint(0, 1) for _ in range(CHROM_LENGTH)]
```

```python
# 2. Fitness function: count number of 1s
def fitness(individual):
    return sum(individual)
```

```python
# 3. Crossover: single-point crossover
def crossover(parent1, parent2):
    point = random.randint(1, CHROM_LENGTH-1)  # crossover point
    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]
    return child1, child2

```

```python
# 4. Mutation: flip bits with MUTATION_RATE probability
def mutate(individual):
    for i in range(CHROM_LENGTH):
        if random.random() < MUTATION_RATE:
            individual[i] = 1 - individual[i]
```

```python
# 5. Initialize population
population = [create_individual() for _ in range(POP_SIZE)]
```

```python
# 6. GA loop
for gen in range(GENERATIONS):
    # Sort population by fitness (descending)
    population.sort(key=fitness, reverse=True)

    # Print best individual in current generation
    best = population[0]
    print(f"Generation {gen+1}: Best = {best}, Fitness = {fitness(best)}")

    # Selection: take top 2 as parents
    parents = population[:2]

    # Create new population
    new_population = parents.copy()  # Keep parents

    # Generate offspring to fill population
    while len(new_population) < POP_SIZE:
        p1, p2 = random.sample(parents, 2)
        child1, child2 = crossover(p1, p2)
        mutate(child1)
        mutate(child2)
        new_population.append(child1)
        if len(new_population) < POP_SIZE:
            new_population.append(child2)

    population = new_population
```
